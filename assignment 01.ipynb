```python
#第二题（2）
def Matrix_multip(M1, M2):
    #设置空变量
    M3 = np.zeros((5, 5), dtype=int)
    #循环
    for i in range(5):
        for j in range(5):
            for k in range(10):
                M3[i, j] = M3[i, j] + M1[i, k] * M2[k, j]
    return M3

print("M1 * M2:")
print(Matrix_multip(M1, M2))
```

```python
#第三题 杨辉三角
def Pascal_triangle(k):
    row = [1]
    #循环公式
    for i in range(k):
        new_row = [1]
        for j in range(len(row) - 1):
            new_row.append(row[j] + row[j + 1])
        new_row.append(1)
        row = new_row

    print(row)

Pascal_triangle(100)
Pascal_triangle(200)
```

```python
#第四题
def Least_moves(x):
    moves = 0
    #判断步数为奇数偶数
    for i in range(x):
        if x == 1:
            break

        if x % 2 == 0:
            x = x // 2
        else:
            x = x - 1

        moves = moves + 1

    print(moves)
Least_moves(2)
Least_moves(5)

import random
x = random.randint(1, 100)  
print("x：", x)
Least_moves(x)
```

```python
#第五题（1）
#引入itertools方便进行排列组合
from itertools import product

def Find_expression(x, show=True):
    count = 0
    #生成组合
    for ops in product(["+", "-", ""], repeat=8):
        expression = "1"
    #组合循环计算
        for i in range(8):
            expression = expression + ops[i] + str(i + 2)
    #计数
        if eval(expression) == x:
            if show:
                print(expression, "=", x)
            count = count + 1

    return count

Find_expression(50)
```

```python
#第五题（2）
#引入matplotlib.pyplot库
import matplotlib.pyplot as plt
#建立空列表
Total_solutions = []
for i in range(1, 101):
    count = Find_expression(i, show=False)
    Total_solutions.append(count)
#画图
plt.plot(range(1, 101), Total_solutions)
plt.xlabel("Target integer")
plt.ylabel("Number of solutions")
plt.title("Total solutions")
plt.grid(True)
plt.show()

print("Maximum:", max(Total_solutions))
print("Minimum:", min(Total_solutions))
#极值
for i in range(100):
    if Total_solutions[i] == max(Total_solutions):
        print("Maximum at:", i + 1)
    if Total_solutions[i] == min(Total_solutions):
        print("Minimum at:", i + 1)
```
